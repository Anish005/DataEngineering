```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
from sklearn.metrics import accuracy_score
```

```python
# ----------------------------
# 1. Generate Dataset
# ----------------------------
np.random.seed(42)
torch.manual_seed(42)

# Training Data
X_train = np.random.rand(1000, 4)
y_train = (np.sum(X_train, axis=1) > 2).astype(int)

# Test Data
X_test = np.random.rand(200, 4)
y_test = (np.sum(X_test, axis=1) > 2).astype(int)

# Convert to PyTorch tensors
X_train = torch.tensor(X_train, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.long)

X_test = torch.tensor(X_test, dtype=torch.float32)
y_test = torch.tensor(y_test, dtype=torch.long)
```

```python
# ----------------------------
# 2. Define Feedforward Neural Network
# ----------------------------
class FeedForwardNN(nn.Module):

    def __init__(self):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(8, 4),
            nn.ReLU(),
            nn.Linear(4, 2)
        )

    def forward(self, x):
        return self.network(x)
```

```python
X_test = np.random.rand(200, 4)
y_test = (np.sum(X_test, axis=1) > 2).astype(int)

X_test = torch.tensor(X_test, dtype=torch.float32)
y_test = torch.tensor(y_test, dtype=torch.long)
```

```python
# Create model
model = FeedForwardNN()

# ----------------------------
# 3. Loss Function & Optimizer
# ----------------------------
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
```

```python
# ----------------------------
# 4. Train the Model
# ----------------------------
epochs = 100

for epoch in range(epochs):

    # Forward pass
    outputs = model(X_train)

    # Calculate loss
    loss = criterion(outputs, y_train)

    # Backpropagation
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 10 == 0:
        print(f"Epoch [{epoch+1}/{epochs}] Loss: {loss.item():.4f}")
```

```python
model.eval()

with torch.no_grad():
    train_outputs = model(X_train)
    train_predictions = torch.argmax(train_outputs, dim=1)

```

```python
correct = (train_predictions == y_train).sum().item()
total = y_train.size(0)

train_accuracy = correct / total

print(f"Training Accuracy: {train_accuracy:.4f}")
```
